# **Intersection Analysis**

This notebook contains the necessary code to identify the genes shared between **TIM3 interactome** and the differentially expressed genes in the **Tumors FA-HNSCC**, and **Tumors HNSCC no FA**.

```python
# Libraries to import

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib_venn import venn2 
from matplotlib_venn import venn3
```

```python
# TIM3 signature genes
# Read genes from a file
tim3_signature = pd.read_csv("signature_tim3.csv")
tim3_signature.head()
```

```python
# Genes in TIM3 signature
tim3_sig = tim3_signature['Gene'].tolist()
print(f"Number of genes in TIM3 signature: {len(tim3_sig)}")
```

```python
# TIM3 interactome
signature = [
    "Krt14","Krt16","Krt5","Ocln","Cdh1","Zeb1","Zeb2","Snai1","Twist1","Twist2","Sox2","Hey1","Fgfr1","Sting","Ptgfr","Ptgs2","Traf1",
    "Traf5","Il33","Vcam1","Ccl2","Eda2r","Cxcl12" 
]

print(len(signature))  
```

**TIM3 interactome genes are present in RNAseq data mouse?**

```python
# RNAseq data mouse
# Read from a file
rnaseq_mouse = pd.read_csv("RNAseq_mouse.csv")
rnaseq_mouse.head()
```

```python
# Show names of columns
rnaseq_mouse.columns
```

```python
# Filter rnaseq_mouse to keep only TIM3 interactome genes (case-insensitive)
tim3_mouse = rnaseq_mouse[rnaseq_mouse['Symbols'].str.upper().isin([gene.upper() for gene in signature])]
print(f"Genes from TIM3 signature found in mouse RNAseq data: {len(tim3_mouse)}")
tim3_mouse
```

```python
# Lossing genes
found = set(tim3_mouse['Symbols'].str.upper())
expected = set(g.upper() for g in signature)

missing = expected - found
print(f"Missing genes: {len(missing)}")
print(missing)
```

```python
# Show columns
rnaseq_mouse.columns
```

```python
# Reorder columns: Symbols, Mouse_FA_PE, then WT controls, then FA experimental
# Exclude IDs column
wt_cols = [col for col in tim3_mouse.columns if col.startswith('Mouse_WT')]
fa_cols = [col for col in tim3_mouse.columns if col.startswith('Mouse_FA') and col != 'Mouse_FA_PE']

new_order = ['Symbols', 'Mouse_FA_PE'] + wt_cols + fa_cols
tim3_mouse_reordered = tim3_mouse[new_order]
save_filename = "genes_in_mouse_RNAseq_mouse.csv"
tim3_mouse_reordered.to_csv(save_filename, index=False)
print(f"Reordered columns: {list(tim3_mouse_reordered.columns)}")
tim3_mouse_reordered
```
